```python
import pandas as pd
import numpy as np
import sqlite3

df = pd.read_csv('../data/processed/Telco-Customer-Churn-Cleaned.csv')
print("Shape:", df.shape)
df.head()
```

```python
# Churn Rate Summary
churn_counts = df['Churn'].value_counts()
churn_rate = (churn_counts['Yes'] / len(df)) * 100
print(f"Baseline Churn Rate: {churn_rate:.2f}%")
```

```python
# Contract Type Churn Breakdown
pd.crosstab(df['Contract'], df['Churn'], normalize='index') * 100
```

```python
# Payment Method Churn Breakdown
pd.crosstab(df['PaymentMethod'], df['Churn'], normalize='index') * 100
```

```python
# Connect to SQLite and execute validation query
conn = sqlite3.connect('../database/churn.db')
q = '''SELECT Contract, COUNT(*) as Total, ROUND(AVG(MonthlyCharges), 2) as AvgCharges FROM telco_churn GROUP BY Contract;'''
pd.read_sql_query(q, conn)
```
